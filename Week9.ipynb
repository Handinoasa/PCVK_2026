```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Dilasi tanpa library
def dilasi_citra(F, w):
    # size image
    p, q = F.shape
    imgD = np.zeros((p, q), dtype=np.uint8)

    # Generate structure element dilasi (SED)
    SED = np.ones((w, w), dtype=np.uint8)
    constant1 = (w - 1) // 2

    # Proses Dilasi
    for i in range(constant1, p - constant1):
        for j in range(constant1, q - constant1):
            temp = F[i - constant1 : i + constant1 + 1, j - constant1 : j + constant1 + 1]
            product = temp * SED
            imgD[i, j] = np.max(product)

    return imgD

# Main program (aligned to the far left, outside the function)
img_d = cv2.imread('/content/drive/MyDrive/Images/plat nomor.jpg', 0)
ret, thresh = cv2.threshold(img_d, 125, 200, cv2.THRESH_BINARY)

w = 5
imgD = dilasi_citra(thresh, w)

plt.figure(figsize=(10, 5))
plt.subplot(131), plt.imshow(img_d, cmap='gray')
plt.title('Citra Awal'), plt.xticks([]), plt.yticks([])
plt.subplot(132), plt.imshow(imgD, cmap='gray')
plt.title('Citra Hasil Dilasi'), plt.xticks([]), plt.yticks([])
plt.show()
```

```python
import cv2 as cv
from matplotlib import pyplot as plt
from skimage import filters
from skimage.morphology import skeletonize
import numpy as np

img = cv.imread('/content/drive/MyDrive/Images/rose_pink.png', 0)

binary = img > filters.threshold_triangle(img)

# true false to one
binary_cp = binary.copy()
binary_cp[binary_cp == True] = 1
binary_cp[binary_cp == False] = 0

# skeletonize image
skeleton = skeletonize(binary_cp)

# print images
plt.figure(figsize=(20, 10))

plt.subplot(121), plt.imshow(img, cmap='gray')
plt.title('Original Image'), plt.xticks([]), plt.yticks([])

plt.subplot(122), plt.imshow(skeleton, cmap='gray')
plt.title('Skeleton Image'), plt.xticks([]), plt.yticks([])
plt.show()
```

```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt

image_path = '/content/drive/MyDrive/Images/ktp1.jpg'
img = cv2.imread(image_path)

if img is None:
    print(f"ERROR: Gambar KTP tidak ditemukan di '{image_path}'. Pastikan nama file sudah benar.")
else:
    # 1. Konversi ke Grayscale
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # 2. Adaptive Thresholding
    thresh = cv2.adaptiveThreshold(
        gray, 255,
        cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
        cv2.THRESH_BINARY_INV,
        15, 5
    )

    # 3. Operasi Morfologi
    kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (2, 2))
    closing = cv2.morphologyEx(thresh, cv2.MORPH_CLOSE, kernel)
    hasil_akhir = cv2.morphologyEx(closing, cv2.MORPH_OPEN, kernel)
    hasil_akhir_inv = cv2.bitwise_not(hasil_akhir)

    # Menampilkan perbandingan proses
    plt.figure(figsize=(15, 10))

    plt.subplot(221), plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.title('1. Citra KTP Asli (RGB)'), plt.xticks([]), plt.yticks([])

    plt.subplot(222), plt.imshow(thresh, cmap='gray')
    plt.title('2. Hasil Adaptive Thresholding'), plt.xticks([]), plt.yticks([])

    plt.subplot(223), plt.imshow(hasil_akhir, cmap='gray')
    plt.title('3. Hasil Closing + Opening (Teks Putih)'), plt.xticks([]), plt.yticks([])

    plt.subplot(224), plt.imshow(hasil_akhir_inv, cmap='gray')
    plt.title('4. Hasil Akhir (Teks Hitam - Siap Ekstraksi)'), plt.xticks([]), plt.yticks([])

    plt.tight_layout()
    plt.show()
```

Agar meningkatkan keterbacaan teks pada citra KTP, pendekatan terbaik adalah menggabungkan Thresholding (Binarisasi) dengan Operasi Morfologi Closing dan Opening. Karena, Foto Scan KTP sering kali memiliki noise berupa pantulan cahaya, stempel, atau latar belakang bertekstur (seperti pola kepulauan Indonesia) yang mengganggu pembacaan teks.

berdasarkan hal tersebut menggunakan filter morfologi berlapis untuk mempertegas goresan huruf (Closing) sekaligus menghilangkan titik-titik noise kecil di area latar belakang (Opening).

Parameter Tambahan yang Diterapkan:
- Adaptive Thresholding (cv2.ADAPTIVE_THRESH_GAUSSIAN_C)
- SE Kotak 2x2 (cv2.MORPH_RECT)
- Kombinasi Sequential (Closing lalu Opening)
