```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('foto.jpg')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
img_rb = img_rgb.copy()
img_rb[:, :, 1] = 0


img_gb = img_rgb.copy()
img_gb[:, :, 0] = 0

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(img_rb)
plt.title("Merah - Biru")

plt.subplot(1,2,2)
plt.imshow(img_gb)
plt.title("Hijau - Biru")
plt.show()
```

```python
# 0: vertikal, 1: horizontal, -1: keduanya
img_flip_v = cv2.flip(img_rgb, 0)
img_flip_h = cv2.flip(img_rgb, 1)
img_flip_vh = cv2.flip(img_rgb, -1)

plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(img_flip_v)
plt.title("Flip Vertikal")

plt.subplot(1,3,2)
plt.imshow(img_flip_h)
plt.title("Flip Horizontal")

plt.subplot(1,3,3)
plt.imshow(img_flip_vh)
plt.title("Flip Keduanya")
plt.show()
```

```python
import cv2
from google.colab.patches import cv2_imshow

def beri_label_dinamis(image_path, text="Dokumentasi PCVK - 2026"):
    img = cv2.imread(image_path)
    h, w = img.shape[:2]

    # Menghitung bar dan font agar proporsional
    bar_height = h // 12
    font_scale = h / 1000  # Skala font berdasarkan tinggi gambar
    thickness = max(1, int(h / 300))

    # Tambahkan bar hitam di bagian paling bawah citra
    img[h-bar_height:h, 0:w] = (0, 0, 0)

    # Meletakkan teks secara dinamis di dalam bar hitam
    font = cv2.FONT_HERSHEY_SIMPLEX
    text_y = h - (bar_height // 3)
    cv2.putText(img, text, (20, text_y), font, fontScale=font_scale, color=(255, 255, 255), thickness=thickness, lineType=cv2.LINE_AA)

    cv2_imshow(img)

# Uji coba
beri_label_dinamis('foto_lanscape.jpg')
beri_label_dinamis('Foto_Potrait.jpg')
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

def buat_thumbnail_persegi(image_path, nama_toko="TOKO PCVK"):
    img = cv2.imread(image_path)
    h, w = img.shape[:2]

    # Tentukan ukuran kanvas persegi (mengikuti sisi terpanjang gambar)
    max_dim = max(h, w)

    # Inisialisasi kanvas putih berukuran max_dim x max_dim
    # Latar putih: isi dengan 255
    canvas = np.full((max_dim, max_dim, 3), 255, dtype=np.uint8)

    # Hitung offset untuk menempatkan citra tepat di tengah (center)
    x_offset = (max_dim - w) // 2
    y_offset = (max_dim - h) // 2

    # Salin citra asli ke area tengah kanvas menggunakan slicing
    canvas[y_offset:y_offset+h, x_offset:x_offset+w] = img

    # Tambahkan Logo/Nama toko (ukurannya proporsional)
    font_scale = max_dim / 800
    thickness = max(1, int(max_dim / 300))
    cv2.putText(canvas, nama_toko, (30, 50), cv2.FONT_HERSHEY_SIMPLEX, fontScale=font_scale, color=(0, 0, 255), thickness=thickness, lineType=cv2.LINE_AA)

    # Opsional: Jika ingin selalu berukuran misal 500x500 untuk web, resize di akhir
    thumbnail_final = cv2.resize(canvas, (500, 500))
    cv2_imshow(thumbnail_final)

# Uji coba
buat_thumbnail_persegi('foto_lanscape.jpg')
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

def buat_thumbnail_persegi_presisi(image_path, nama_toko="TOKO PCVK"):
    img = cv2.imread(image_path)
    h, w = img.shape[:2]

    max_dim = max(h, w)
    padding = int(max_dim * 0.10)
    canvas_size = max_dim + (2 * padding)
    canvas = np.full((canvas_size, canvas_size, 3), 255, dtype=np.uint8)
    x_offset = (canvas_size - w) // 2
    y_offset = (canvas_size - h) // 2

    canvas[y_offset:y_offset+h, x_offset:x_offset+w] = img
    font = cv2.FONT_HERSHEY_SIMPLEX
    font_scale = max_dim / 600
    thickness = max(1, int(max_dim / 200))


    (text_w, text_h), baseline = cv2.getTextSize(nama_toko, font, font_scale, thickness)
    margin = int(max_dim * 0.02)
    text_x = margin + 15
    text_y = margin + text_h + 15

    pt1 = (margin, margin)
    pt2 = (margin + text_w + 30, margin + text_h + 20 + baseline)
    cv2.rectangle(canvas, pt1, pt2, (255, 255, 255), -1)
    cv2.putText(canvas, nama_toko, (text_x, text_y), font, fontScale=font_scale, color=(0, 0, 255), thickness=thickness, lineType=cv2.LINE_AA)
    thumbnail_final = cv2.resize(canvas, (500, 500))
    cv2_imshow(thumbnail_final)


buat_thumbnail_persegi_presisi('Foto_Potrait.jpg')
```
